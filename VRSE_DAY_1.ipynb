```python
def isAnagram(s,t):
    if(len(s) != len(t)):
        return False
    freq = [0]*26
    for i in range(len(s)):
        sIdx = ord(s[i])-ord('a')
        freq[sIdx]+=1

        tIdx = ord(t[i])-ord('a')
        freq[tIdx]-=1

    for i in range(26):
        if(freq[i] != 0):
            return False
    return True
```

```python
def removeDuplicates(arr):
    res = []
    for i in arr:
        if i not in res:
            res.append(i)
    return res

res = removeDuplicates([1,1,2,2,2,3,3,3,5,5,5,4])
print(res)
```

```python
def findSingles(arr):
    freq = dict()
    for i in arr:
        if(freq.get(i) == None):
            freq[i] = 1
        else:
            freq[i] += 1
    res = []
    for i in freq.keys():
        if(freq[i] == 1):
            res.append(i)
    return res
res = findSingles([1,1,2,5,4,7,8,2,1,4,5,6,3,7])
print(res)
```
